```python
#Multiplication Table
n=int(input())
for i in range (1,11):
    print(n*i, end=" ")
```

```python
#Even positioned characters
def utility(s):
    for i in range(0,len(s),2):
        print(s[i], end="")
```

```python
s="DoctorPhemomenal"
utility(s)
```

```python
#While loop in python
def printInDecreasing(x):
    while x>=0:
        print(x,end=" ")
        x-=1
```

```python
x=5
printInDecreasing(x)
```

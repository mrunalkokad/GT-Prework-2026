```python
#Even positioned characters
def utility(s):
    for i in range(0,len(s),2):
        print(s[i], end="")
```

```python
s="DoctorPhemomenal"
utility(s)
```

```python
#While loop in python
def printInDecreasing(x):
    while x>=0:
        print(x,end=" ")
        x-=1
```

```python
x=5
printInDecreasing(x)
```

```python
#Jumping through while
def PrintIncreasingPower(x):
    i=1
    while i*i<=x:
        print(i*i, end=" ")
        i+=1
```

```python
x=20
PrintIncreasingPower(x)
```

```python
#Zero Converter
def pos(n):
    while n>0:
        n -=1
        print(n,end=" ")

def neg(n):
    while n<0:
        print(n,end=" ")
        n +=1
    print(0,end=" ")
```

```python
n=-3
neg(n)
```

```python
#Leap year
class Solution:
    def checkyear(self,n):
        if n%400==0:
            return True
        elif n%100==0:
            return False
        elif n%4==0:
            return True
        else:
            return False
```

```python
obj=Solution()
print(obj.checkyear(2019))
```

```python
#Calculator
class Solution:
    def calculate(self, a: int, b: int, operator: int) -> None:
        if operator == 1:
            print(a + b, end="")
        elif operator == 2:
            print(a - b, end="")
        elif operator == 3:
            print(a * b, end="")
        else:
            print("Invalid Input", end="")
```

```python
obj=Solution()
obj.calculate(10,5,1)
```

```python
#Table Difference
def difference (n1,n2):
    for i in range(1,11):
        print((n1*i)-(n2*i),end=" ")
```

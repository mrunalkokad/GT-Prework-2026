```python
#Function with return
class Solution:
    def returnValueFunction(self, n):
        return 2*n
```

```python
obj=Solution()
obj.returnValueFunction(2)
```

```python
#Function with no arguments
class Solution:
    def HelloFunction(self):
        print("Hello")
```

```python
obj=Solution()
obj.HelloFunction()
```

```python
#Function with arguements
class Solution:
    def arguementFunction(self,a,b):
        return a+b
```

```python
obj=Solution()
obj.arguementFunction(5,6)
```

```python
#First digit of a number
def firstdigit(n):
    while n>=10:
        n=n//10
    return n
```

```python
firstdigit(123)
```

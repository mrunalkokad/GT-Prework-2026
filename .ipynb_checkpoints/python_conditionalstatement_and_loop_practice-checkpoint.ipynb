```python
#Even positioned characters
def utility(s):
    for i in range(0,len(s),2):
        print(s[i], end="")
```

```python
#While loop in python
def printInDecreasing(x):
    while x>=0:
        print(x,end=" ")
        x-=1
```

```python
x=5
printInDecreasing(x)
```

```python
#Jumping through while
def PrintIncreasingPower(x):
    i=1
    while i*i<=x:
        print(i*i, end=" ")
        i+=1
```

```python
x=20
PrintIncreasingPower(x)
```

```python
#Zero Converter
def pos(n):
    while n>0:
        n -=1
        print(n,end=" ")

def neg(n):
    while n<0:
        print(n,end=" ")
        n +=1
    print(0,end=" ")
```

```python
n=-3
neg(n)
```

```python
#Leap year
class Solution:
    def checkyear(self,n):
        if n%400==0:
            return True
        elif n%100==0:
            return False
        elif n%4==0:
            return True
        else:
            return False
```

```python
obj=Solution()
print(obj.checkyear(2019))
```

```python
#Calculator
class Solution:
    def calculate(self, a: int, b: int, operator: int) -> None:
        if operator == 1:
            print(a + b, end="")
        elif operator == 2:
            print(a - b, end="")
        elif operator == 3:
            print(a * b, end="")
        else:
            print("Invalid Input", end="")
```

```python
obj=Solution()
obj.calculate(10,5,1)
```

```python
#Right Angle Triangle 2
class Solution:
    def printPattern(self, n):
        for i in range(1,n+1):
            if i==1:
                print("*")
            elif i==n:
                print("* "*n)
            else:
                print("*"+" "*(2*i-3)+"*")
```

```python
obj=Solution()
obj.printPattern(5)
```

```python
#Factorial
class Solution:
    def factorial(self,n:int)->int:
        fact=1

        for i in range (1,n+1):
            fact=fact*i
        return fact
```

```python
obj=Solution()
obj.factorial(5)
```

```python
#All divisors of a number
class Solution:
    def getDivisors(self,n):
        ans=[]

        for i in range(1,int(n**0.5)+1):
            if n%i==0:
                ans.append(i)

                if i !=n//i:
                    ans.append(n//i)
        ans.sort()
        return ans
```

```python
obj=Solution()
obj.getDivisors(20)
```

```python
#Prime Number
class Solution:
    def isPrime(self,n):
        if n<=1:
            return False
        for i in range(2,int(n**0.5)+1):
            if n%i==0:
                return False
        return True
```

```python
obj=Solution()
obj.isPrime(7)
```

```python
#Next prime number
def nextPrime(n):
    while True:
        n += 1
        print("checking", n)   # temporary debug line
        prime = True
        if n <= 1:
            prime = False
        else:
            for i in range(2, int(n**0.5) + 1):
                if n % i == 0:
                    prime = False
                    break
        if prime:
            return n
print(nextPrime(17))
```

```python
#Fibonacci number
class Solution:
    def nthFibonacci(self, n: int) -> int:
        if n == 0:
            return 0
        elif n == 1:
            return 1

        a = 0
        b = 1

        for i in range(2, n + 1):
            c = a + b
            a = b
            b = c

        return b
```

```python
obj=Solution()
obj.nthFibonacci(5)
```

```python
#GCD of two numbers
class Solution:
    def gcd(self,a,b):
        while b!=0:
            a,b=b,a%b
        return a
```

```python
obj=Solution()
obj.gcd(20,28)
```

```python
#LCM of two numbers
class Solution:

    def lcm(self, a, b):
        x = a
        y = b

        while y != 0:
            x, y = y, x % y

        return (a * b) // x
```

```python
obj=Solution()
obj.lcm(42,8)
```

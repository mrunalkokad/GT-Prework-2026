```python
#Multiplication Table
n=int(input())
for i in range (1,11):
    print(n*i, end=" ")
```

```python
#Even positioned characters
def utility(s):
    for i in range(0,len(s),2):
        print(s[i], end="")
```

```python
s="DoctorPhemomenal"
utility(s)
```

```python
#While loop in python
def printInDecreasing(x):
    while x>=0:
        print(x,end=" ")
        x-=1
```

```python
x=5
printInDecreasing(x)
```

```python
#Jumping through while
def PrintIncreasingPower(x):
    i=1
    while i*i<=x:
        print(i*i, end=" ")
        i+=1
```

```python
x=20
PrintIncreasingPower(x)
```

```python
#Zero Converter
def pos(n):
    while n>0:
        n -=1
        print(n,end=" ")

def neg(n):
    while n<0:
        print(n,end=" ")
        n +=1
    print(0,end=" ")
```

```python
n=-3
neg(n)
```

```python
#THe else statement
a=int(input())
if a>100:
    print("Big")
else:
    print("Small")
```

```python
#The fizzbuzz program
n=int(input())
if n%3==0 and n%5==0:
    print("Fizzbuzz")
elif n%3==0:
    print("Fizz")
elif n%5==0:
    print("Buzz")
else:
    print(n)
```
